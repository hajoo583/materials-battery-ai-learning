```python
import re

email = input("What is ur email? ")

if re.search(".*@.*", email):
    print("Valid")
else:
    print("Invalid")
```

```python
import re

email = input("What is ur email? ")

if re.search(r"^\w+@\w+\.(edu|com|gov)$", email):
    print("Valid")
else:
    print("Invalid")
```

```python
import re

email = input("What is ur email? ")

if re.search(r"^\w+@(\w + \.)?\w+\.(edu|com|gov)$", email, re.IGNORECASE):
    print("Valid")
else:
    print("Invalid")
```

```python
import re

name = input("What is your name? ").strip()

matches = re.search(r"^(.+), (.+)$", name)
if matches:
    last, first = matches.groups()
    name = f"{first} {last}"
print(f"Hello, {name}")
```

```python
import re

name = input("What is your name? ").strip()

matches = re.search(r"^(.+), *(.+)$", name)
if matches:
    last, first = matches.groups()
    name = f"{first} {last}"
print(f"Hello, {name}")
```

```python
import re

url = input("URL: ").strip()


if matches := re.search(r"^(https?://)?(?:www\.)?twitter\.com/.+$", "", url, re.IGNORECASE):
    print(f"Username:", matches.group(1) )
```

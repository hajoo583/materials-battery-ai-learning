```python
class Student:
    def __init__(self, name, house, patronus):
        self.name = name
        self.house = house
        self.patronus = patronus

    def __str__(self):
        return f"{self.name} from {self.house}"
    
    def charm(self):
        match self.patronus:
            case "Stag":
                return "🦌"
            case "Otter":
                return "🦦"
            case "Jack Russell terrier":
                return "🐶"
            case _:
                return "🪄"

def main():
    student = get_student()
    print(student)
    print(student.charm())

def get_student():
    name = input("Name= ")
    house = input("House= ")
    patronus = input("patronus= ")
    student = Student(name, house, patronus)
    return student 


if __name__ == "__main__":
    main()
    
```

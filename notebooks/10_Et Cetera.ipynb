```python
students = [
    {"name": "Hajoo", "house": "a"},
    {"name": "John", "house": "b"}


]

houses = []
for student in students:
    if student["house"] not in houses:
        houses.append(student["house"])

for house in sorted(houses):
    print(house)
```

```python
# class set()
students = [
    {"name": "Hajoo", "house": "a"},
    {"name": "John", "house": "b"}


]

houses = set()
for student in students:
        houses.add(student["house"])

for house in sorted(houses):
    print(house)
```

```python
voltage_v = 3.7
capacity_ah = 2.5
cycle_count = 100
battery_name = "Example Cell"

energy_wh = voltage_v * capacity_ah

print("Battery:", battery_name)
print("Voltage:", voltage_v, "V")
print("Capacity:", capacity_ah, "Ah")
print("Energy:", energy_wh, "Wh")
print("Cycle count:", cycle_count)
```

```python
print(type(voltage_v))
print(type(capacity_ah))
print(type(cycle_count))
print(type(battery_name))
```
